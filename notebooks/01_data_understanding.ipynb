# 01 — Data Understanding & Raw Dataset Profiling

**Project:** E-commerce Revenue & Customer Retention Intelligence  
**Dataset:** Olist Brazilian E-Commerce Public Dataset (9 CSV tables, `1,550,922` total rows)  
**Immutability Guarantee:** Raw files in `data/raw/` are read-only and verified via SHA-256 hashes.

### Key Grain Findings
- `orders` (`99,441` rows) and `customers` (`99,441` rows) share a 1:1 relationship on `customer_id`, whereas **`customer_unique_id` (`96,096` unique)** represents the actual customer.
- `order_items` (`112,650` rows), `order_payments` (`103,886` rows), and `order_reviews` (`99,224` rows) have **different grains** and must be aggregated to `order_id` before joining.

```python
import json
from pathlib import Path
import pandas as pd

ROOT = Path("..").resolve()
meta = json.loads((ROOT / "reports" / "dataset_metadata.json").read_text(encoding="utf-8"))

rows = []
for fname, p in meta["profiles"].items():
    rows.append({
        "file": fname,
        "rows": p["rows"],
        "cols": p["columns"],
        "duplicate_rows": p["duplicate_rows"],
        "missing_cells": p["total_missing_cells"],
        "missing_pct": p["missing_pct_overall"],
    })
summary_df = pd.DataFrame(rows)
print(summary_df.to_string(index=False))
```
